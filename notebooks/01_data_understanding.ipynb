```python
import pandas as pd
import numpy as np
import os
from pathlib import Path

print("Libraries imported successfully!")
```

```python
DATA_PATH = Path("../data/raw")

csv_files = list(DATA_PATH.glob("*.csv"))

print("Total CSV files:", len(csv_files))

for file in csv_files:
    print(file.name)
```

```python
customers = pd.read_csv(
    DATA_PATH / "olist_customers_dataset.csv"
)

customers.head()
```

```python
print("Total rows:", customers.shape[0])
print("Total columns:", customers.shape[1])

print("\nColumn names:")
print(customers.columns.tolist())
```

```python
print("Dataset Information:")
customers.info()

print("\nMissing Values:")
print(customers.isnull().sum())

print("\nDuplicate Rows:")
print(customers.duplicated().sum())
```

```python
# Load orders dataset

orders = pd.read_csv(
    DATA_PATH / "olist_orders_dataset.csv"
)

# Display first five records

orders.head()
```

```python
# Dataset dimensions

print("Total rows:", orders.shape[0])
print("Total columns:", orders.shape[1])

print("\nColumn names:")
print(orders.columns.tolist())
```

```python
print("Dataset Information:")

orders.info()

print("\nMissing Values:")

print(orders.isnull().sum())

print("\nDuplicate Rows:")

print(orders.duplicated().sum())
```

```python
# Count orders by status

orders["order_status"].value_counts()
```

```python
# Merge customers and orders

customer_orders = pd.merge(
    customers,
    orders,
    on="customer_id",
    how="inner"
)

customer_orders.head()
```

```python
print("Merged dataset shape:", customer_orders.shape)

print("\nColumns:")
print(customer_orders.columns.tolist())

print("\nTotal orders:", customer_orders["order_id"].nunique())

print("\nUnique customers:",
      customer_orders["customer_unique_id"].nunique())
```

```python
status_analysis = (
    customer_orders["order_status"]
    .value_counts()
    .reset_index()
)

status_analysis.columns = ["Order Status", "Count"]

status_analysis
```

```python
top_cities = (
    customers["customer_city"]
    .value_counts()
    .head(10)
)

top_cities
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(12, 6))

sns.barplot(
    x=top_cities.values,
    y=top_cities.index
)

plt.title("Top 10 Cities by Customer Records")
plt.xlabel("Number of Customer Records")
plt.ylabel("City")

plt.show()
```

```python
# Count orders placed by each unique customer

orders_per_customer = (
    customer_orders
    .groupby("customer_unique_id")["order_id"]
    .nunique()
)

orders_per_customer.head(10)
```

```python
# Customers who placed more than one order

repeat_customers = orders_per_customer[
    orders_per_customer > 1
]

one_time_customers = orders_per_customer[
    orders_per_customer == 1
]

print("Total unique customers:",
      len(orders_per_customer))

print("Repeat customers:",
      len(repeat_customers))

print("One-time customers:",
      len(one_time_customers))
```

```python
repeat_purchase_rate = (
    len(repeat_customers) /
    len(orders_per_customer)
) * 100

print(
    f"Repeat Purchase Rate: "
    f"{repeat_purchase_rate:.2f}%"
)
```

```python
unique_customer_data = (
    customers
    .drop_duplicates(
        subset="customer_unique_id",
        keep="first"
    )
)

top_unique_customer_cities = (
    unique_customer_data
    .groupby("customer_city")["customer_unique_id"]
    .nunique()
    .nlargest(10)
)

top_unique_customer_cities
```

```python
# Distribution of orders per customer

order_frequency = (
    orders_per_customer
    .value_counts()
    .sort_index()
)

print(order_frequency)
```

```python
plt.figure(figsize=(10, 5))

sns.barplot(
    x=order_frequency.index.astype(str),
    y=order_frequency.values
)

plt.title("Customer Order Frequency Distribution")
plt.xlabel("Number of Orders")
plt.ylabel("Number of Customers")

plt.show()
```

```python
average_orders = orders_per_customer.mean()

median_orders = orders_per_customer.median()

print("Average orders per customer:", round(average_orders, 2))
print("Median orders per customer:", median_orders)
```
